```python
import numpy as np
import scipy
import soundfile as sf
import matplotlib.pyplot as plt
from IPython.display import Audio, display
print('numpy', np.__version__, '| scipy', scipy.__version__, '| all imports OK')
```

```python
noisy, fs = sf.read('test_noisy.wav')
clean, _ = sf.read('outputs/test_denoised.wav')
print('before'); display(Audio(noisy, rate=fs))
print('after RNNoise'); display(Audio(clean, rate=fs))
```

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 5), sharex=True)
for a, sig, name in zip(ax, [noisy, clean], ['noisy', 'RNNoise']):
    a.specgram(sig, NFFT=960, Fs=fs, noverlap=480, cmap='magma')
    a.set_title(name); a.set_ylabel('Hz'); a.set_ylim(0, 8000)
ax[-1].set_xlabel('seconds'); plt.tight_layout()
```
